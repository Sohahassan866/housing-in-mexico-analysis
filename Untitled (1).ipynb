```python
# Import Matplotlib, pandas, and plotly
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
```

```python
df1 = pd.read_csv("data/brasil-real-estate-1.csv")
df1.head()
```

```python
df1.head()
df1.info()
```

```python
df1.dropna(inplace=True)
df1.head(20)
```

```python
df1[["lat", "lon"]] = df1["lat-lon"].str.split(",", expand=True) .astype(float)
df1
```

```python
df1["state"] = df1["place_with_parent_names"].str.split('|', expand=True)[2]
```

```python
df1["price_usd"] = (df1["price_usd"].str.replace("$", "", regex=False).str.replace(",", "").astype(float))
```

```python
df1.drop(columns=[
    "place_with_parent_names","lat-lon"], inplace=True)
df1.info()
```

```python
df2 = pd.read_csv("data/brasil-real-estate-2.csv")
```

```python
df2['price_usd'] = df2['price_brl'] / 3.19
df2.info
```

```python
df2.dropna(inplace=True)
df2.drop(columns=["price_brl"], inplace=True)
```

```python
df= pd.concat([df1, df2])
print("df shape:", df.shape)
```

```python
fig = px.scatter_mapbox(
   df,
    lat="lat",
    lon="lon",
    center={"lat": -14.2, "lon": -51.9},  # Map will be centered on Brazil
    width=600,
    height=600,
    hover_data=["price_usd"],  # Display price when hovering mouse over house
)

fig.update_layout(mapbox_style="open-street-map")

fig.show()
```

```python
# Create summary statistics for selected columns
summary_stats = df[["area_m2", "price_usd"]].describe()

# Display the result
print(summary_stats)
```

```python
# Build histogram
plt.hist(df["price_usd"].head(20000))


# Label axes
plt.xlabel("Price [USD]")

# Add title
plt.title("Distribution of Home Prices")
```

```python
# Build box plot
plt.boxplot(df["area_m2"].head(20000))


# Label x-axis

plt.xlabel("Area [sq meters]")
# Add title
plt.title("Distribution of Home Sizes")
```

```python
mean_price_by_region = df.groupby("region")["price_usd"].mean().sort_values()
mean_price_by_region 
```

```python
mean_price_by_region.plot(
    kind="bar",
    xlabel="Region",
    ylabel="Mean Price [USD]",
    title="Mean Home Price by Region",
);
plt.savefig("images/1-5-15.png",dpi=150)
```

```python
df_south = df[df["region"] == "South"].copy()
df_south.head()
```

```python
homes_by_state = df_south['state'].value_counts()
homes_by_state
```

```python
# Subset data
df_south_rgs = df_south[df_south["state"] == "Rio Grande do Sul"] 
plt.scatter(df_south_rgs["area_m2"], df_south_rgs["price_usd"])

# label axes
plt.xlabel("Area [sq meters]")
plt.ylabel("Price [USD]")

# Add title
plt.title("Rio Grande do Sul: Price vs. Area")
```

```python
south_states_corr = {}

# Filter the South region
df_south = df[df["region"] == "South"]

# Loop through each state in the South region
for state in df_south["state"].unique():
    df_state = df_south[df_south["state"] == state]
    corr = df_state["area_m2"].corr(df_state["price_usd"])
    south_states_corr[state] = corr

# Display the result
south_states_corr
```
